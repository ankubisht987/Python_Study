```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
print("=== Zomato Sales Analysis ===")
print("\n Row & Columns: " , sales_data.shape)
print("\n First 3 Restaurant Data: \n", sales_data[:3])
```

```python
## Total Sales Per Year

print("\n Total Sales Per year: ", sales_data[:, 1:].sum(axis=0) )
## Yearly Total
print("\n Yearly Total 2021 - 2024 which has id 2: ", sales_data[1:2,0:])
```

```python
## Minimum Sales Per restaurant
## Maximum Sales Per Restaurant
perRestaurant_minSales =  sales_data[:,1:].min(axis=1)
perRestaurant_maxSales =  sales_data[:,1:].max(axis=0)
minIndex = np.argmin(perRestaurant_minSales)
maxIndex = np.argmax(perRestaurant_maxSales)
print("Minimum Sales Per Restaurant is: ", perRestaurant_minSales)
print("Maximum Sales Per Restaurant is: ", perRestaurant_maxSales)
print("Minimum Sales One Restaurant is: ", np.min(perRestaurant_minSales) ," at Index: ", minIndex)
print("Maximum Sales One Restaurant is: ", np.max(perRestaurant_maxSales) ," at Index: ", maxIndex)

### Average Sales Per Restaurant
Col_avg_sales = np.mean(sales_data[:,1:], axis=0)
print("Average Of Column: ", Col_avg_sales)
```

```python
cumsum = np.cumsum(sales_data[:,1:],axis=1)
print(cumsum)
```

### Graphs

```python
plt.figure(figsize=(10,6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average Cumulative Sales across All Restaurant ")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)
plt.show()  
```
